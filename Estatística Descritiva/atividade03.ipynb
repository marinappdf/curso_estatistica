**Digamos que eu queira analisar apenas os valores de venda entre 1000 e 2000**
**Para facilitar o código, vou salvar a coluna sales em uma variável do tipo series**

**A distribuição dos valores de venda é:**

````python
plt.figure(figsize=(12, 6))
plt.hist(
    serie.dropna(),
    bins=50,
    edgecolor="black",
    alpha=0.7)
plt.grid(False) 
````

#### O nome disso é HISTROGRAMA

[Mais um gif legal](https://iconscout.com/pt/lottie-animation/grafico-de-barras-multiplas-animation_12720731)

#### Para organizar o código, vou salvar valores em variáveis, e vou associar uma cor a cada um deles

````python
estatisticas = {
    "Min": serie.min(),
    "Q1": serie.quantile(q=0.25),
    "Mediana": serie.quantile(q=0.50),
    "Mean": serie.mean(),
    "Q3": serie.quantile(q=0.75),
    "Max": serie.max()
}
````

````python

# Linhas verticais pontilhadas
cores = {
    "Min": "red",
    "Q1": "orange",
    "Mediana": "green",
    "Mean": "blue",
    "Q3": "purple",
    "Max": "black"
}
````

````python

for nome, valor in estatisticas.items():
    plt.axvline(
        valor,
        color=cores[nome],
        linestyle=":",
        linewidth=2,
        label=f"{nome}: {valor:.2f}"
    )
plt.legend()
plt.tight_layout()
````

### Agora vamos ver tudo junto

````python
plt.figure(figsize=(12, 6))
plt.hist(
    serie.dropna(),
    bins=50,
    edgecolor="black",
    alpha=0.7)
plt.grid(False) 

for nome, valor in estatisticas.items():
    plt.axvline(
        valor,
        color=cores[nome],
        linestyle=":",
        linewidth=2,
        label=f"{nome}: {valor:.2f}"
    )
plt.legend()
plt.tight_layout()

plt.xlabel("Valor")
plt.ylabel("Frequência")
plt.title("Histograma com Estatísticas")


plt.show()
````
