### <div style="color:red; text-align:center"> Estatística descritiva != Estatística Inferencial

|  <h2 style="text-align:center"> Descritiva |  <h2 style="text-align:center"> Inferencial|
|---|---|
| <h3 style="text-align:center"> Descrever os dados |  <h3 style="text-align:center">Inferência de padrões|
| <h3 style="text-align:center"> Propriedades exatas | <h3 style="text-align:center"> Conclusões, não certezas|
|  <h3 style="text-align:center">Objetivo: organizar, descrever e resumir um conjunto de dados | <h3 style="text-align:center"> Objetivo: fazer afirmações a partir de um conjunto de valores representativo (amostra) sobre um universo (população)|

</br>
</br>
</div>


#### Primeiro, separamos uma amostra da população que queremos estudar. Por exemplo, se quero fazer uma pesquisa sobre a intenção de votos, eu não preciso entrevistar todas as pessoas, posso escolher uma amostra da população e realizar a pesquisa com eles.
#### Em seguida, após a limpeza e transformação dos dados, fazemos a análise descritiva, vemos os valores estatísticos que descrevem a amostra.
#### Com isso, vamos escolher o modelo estatístico que melhor descreve o comportamento da amostra. O modelo estatísco é uma equação matemática que replica o comportamento da amostra, é ela que vai nos permitir expandir as previsões da amostra para toda a população
#### Simultâneamente, ao longo de todos esses processos, podemos criar suposições, hipóteses, sobre a população toda, para além da amostra.
#### Aplicando testes de hipóteses e outros teste estatísticos, podemos verificar se as hipóteses que fizemos são possíveis, são razoáveis.
#### Com isso, podemos tirar conclusões do estudo. Essas etapas não são sequênciais, elas ocorrem simultâneamente, num fluxo de ida e vinda.

##### A diferença entre moda, mediana e média é exemplificada na imagem abaixo. 
- Moda é o ponto mais alto da curva, porque é o ponto com maior contagem. Ou seja, moda é o valor mais presente.
- Mediana é o valor central, que divide os dados pela metade. Ou seja, com 50% dos pontos ficam de um lado, e 50% ficam do outro.
- Já a média é o valor que os dados teriam se todos tivessem a mesma frequência.

Desvio padrão e Variância
</h2>
O quanto os pontos variam? Qual a distância entre eles?

Para repsonder isso, uma ideia é medir a média da disância entre cada ponto e a média.

Por exemplo, considerando os dados [1, 0, 6, 9], a média será igual a

`Média = (1 + 0 + 6 + 9)/4`

`Média = 16/4`

`Média = 4`

a distância para cada um é:

```
0 - 4 = -4
1 - 4 = -3
6 - 4 = 2
9 - 4 = 5
```

A média desse valor é 


`Média da variação = (-4 -3 +2 +5)/4`


`Média da Variação = 0`


. . . ué?


Então, essa conta nos leva de volta para média, e com um valor negativo!
Para fugir disso, vamos elevar cada diferença ao quadrado, assim o negativo vai embora, e vamo divir por n, o número de ponto. Isso é a variância.

`Variância = ( (-4)*(-4) + (-3)*(-3) + (2)*(2) + (5)*(5) ) / 4`

`Variância = ( 16 + 9 + 4 + 25 ) / 4`

`Variância = (54) / 4`

`Variância = 13,5`

Mas . . . o que isso significa? O quadrado não aumentou o valor das distâncias?

Sim! Por isso que o desvio padrão é a raiz da variância!

`Desvio Padrão = raizQuadrada( Variância )`

`Desvio Padrão = raizQuadrada( 13,5 )`

`Desvio Padrão = 3,67`

##### A interpretação é: os valores se distanciam da média, em média, 5,4 unidades.

##### Isso indica o quão longe, em geral, os seus valores se encontram do valor esperado.

##### Por essa relação entre desvio padrão e variância, elas são conhecidas como:
 
##### Desvio padrão:  σ

##### Variância: σ2

- Obs 1: a média pode ser chamada de  μ
- Obs 2: para outros modelos estatísticos e outras distribuições, existem diferentes formas de calcular a média e o desvio padrão.

##### 4) Faça um histograma e marque os valores calculados na questão anterior, para a coluna `unit_price`:

Considere o conjunto de dados disponível [neste link](https://www.kaggle.com/datasets/abbas829/ecommerce-sales-dataset). Baixe os dados e use pandas para calcular.

**Dicas de comandos**:

````python
serie = df['unit_price']
````

````python
estatisticas = {
    "Min": serie.min(),
    "Q1": serie.quantile(q=0.25),
    "Mediana": serie.quantile(q=0.50),
    "Mean": serie.mean(),
    "Q3": serie.quantile(q=0.75),
    "Max": serie.max()
}
````

````python
# Linhas verticais pontilhadas
cores = {
    "Min": "red",
    "Q1": "orange",
    "Mediana": "green",
    "Mean": "blue",
    "Q3": "purple",
    "Max": "black"
}
````

````python
plt.axvline(
    serie.quantile(q=0.25),
    color="red",
    linestyle=":",
    linewidth=2,
    label=f"25%: {serie.quantile(q=0.25):.2f}"
)

````

````python
for nome, valor in estatisticas.items():
    plt.axvline(
        valor,
        color=cores[nome],
        linestyle=":",
        linewidth=2,
        label=f"{nome}: {valor:.2f}"
    )
plt.legend()
plt.tight_layout()
````

### Agora vamos ver tudo junto

````python
plt.figure(figsize=(12, 6))
plt.hist(
    serie.dropna(),
    bins=50,
    edgecolor="black",
    alpha=0.7)
plt.grid(False) 

for nome, valor in estatisticas.items():
    plt.axvline(
        valor,
        color=cores[nome],
        linestyle=":",
        linewidth=2,
        label=f"{nome}: {valor:.2f}"
    )
plt.legend()
plt.tight_layout()

plt.xlabel("Valor")
plt.ylabel("Frequência")
plt.title("Histograma com Estatísticas")


plt.show()
````
