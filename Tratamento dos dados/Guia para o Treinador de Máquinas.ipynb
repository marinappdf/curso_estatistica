## Limpeza dos dados: Identificar nulos e faltantes

Valores marcados como NA, None, Vazio, etc. Eles irão prejudicar a análise e precisam ser corrigidos.

As estratégias para resolver esse problema podem ser resumidas em: ***deletar*** ou ***substituir***.

|Ação | Prós| Contras | Algorítimo|
|---|---|---|---|
| Deletar todas as linhas com pelo menos 1 valor nulo | Fácil, rápido, não altera o comportamento da amostra| Para bases pequenas, ou para muitos valores nulos, deletar pode prejudicar o tamanho da amostra, e leva a perda de valores não nulos.| Filtrar e deletar: `df.dropna()`|
| Substituir pela média/moda| Não há perda de dados, não altera o comportamento médio| Altera o comportamento local, pode afetar a análise| Filtrar e substituir pela média/moda da coluna: `df.fillna()` e `df.mean()`. Para dados categóricos, usar o valor mais frequente.|
| Substituir por uma nova categoria "Não informado", ou 0 | Não há perda de dados | Pode alterar o comportamento médio e local, adiciona mais uma categoria para análise| Filtrar e substituir: `df.fillna()`e `= "Não informado"`| 
|Deletar a coluna: quando há uma coluna com muitos valores nulos, pode ser vantajoso excluir ela da análise| Fácil, não interfere no comportamento geral | Se a variável for relevante para a análise, pode prejudicar o resultado| Deletar: `df.drop()`|

#### Para organizar o código, vou salvar valores em variáveis, e vou associar uma cor a cada um deles

````python
estatisticas = {
    "Min": serie.min(),
    "Q1": serie.quantile(q=0.25),
    "Mediana": serie.quantile(q=0.50),
    "Mean": serie.mean(),
    "Q3": serie.quantile(q=0.75),
    "Max": serie.max()
}
````

````python

# Linhas verticais pontilhadas
cores = {
    "Min": "red",
    "Q1": "orange",
    "Mediana": "green",
    "Mean": "blue",
    "Q3": "purple",
    "Max": "black"
}
````
